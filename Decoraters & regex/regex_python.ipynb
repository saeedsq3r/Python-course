```python
import re
```

```python
'''
                Meta Characters
1. \ Used to drop the special meaning of character followingit.
2. [] Tepresent a character class.
3. ^ Matches the beginning.
4. $ Matches the end.
5. . Matches any character except newline.
6. | Mean OR(Matches with any of the character) separated by it.
7. ? Matches zero or one occurrence.
8. * Any number of occurrences (including 0 occurrences).
9. + One or more occurrences.
10. {} Indicate the numer of occurrences of a preceding regex to match.
11. () Enclose a group of Regex.
'''
```

## **Regex Sets**

```python
import re

a = 'charli and the chocolate factory'
b = '12hi34jhon#$@^67hello73,89HELLO'
```

```python
match = re.findall("[atx]",a)
match = re.findall("[^atx]",a)
match = re.findall("[a-t]",a)
match = re.findall("[1234]",b)
match = re.findall("[0-9]",b)
match = re.findall("[0-7][0-9]",b)
match = re.findall("[a-zA-Z]",b)
match = re.findall("[$]",b)
print(match)
```

## **Functions in Regex**

```python
# re.findall()
a = """Jhon has scored 89 marks
Lisa has scored 90 marks
David has scored 70 marks
"""

match = re.findall(r"\d+",a)
match = re.findall(r"[A-Z][a-z]*",a)
print(match)
```

```python
# re.compile()
p = re.compile(r"[a-d]")
p = re.compile(r"[0-9]")
p = re.compile(r"\d+")
print(re.findall(p,a))
```

```python
# re.split()
b = "fantatstics 5 turtles"
print(re.split("\d+",b))
```

```python
# re.sub()
print(re.sub("\s+","",a))
print(re.sub("\s+","_",a))
```

```python
# re.subn()
print(re.subn("\s+","",a))
print(re.subn("\s+","_",a))
```

```python
# re.escape()
print(re.escape(a))
```

## **Match Object in RegeEx**

```python
import re

a = "John has score 98 marks"
```

```python
match = re.search("\d+",a)
print(match)
print(match.re)
print(match.string)
print(match.start())
print(match.end())
print(match.span())
print(match.group())
```

```python
match = re.search("\w{2} s",a)
print(match)
print(match.re)
print(match.string)
print(match.start())
print(match.end())
print(match.span())
print(match.group())
```

### **Phone Number and Email Verification and Web Scrapping Using RegEx**

```python
import re
phn = "222-444-8787"
```

```python
if re.search("\d{3}-\d{3}-\d{4}",phn):
    print("varified")
else:
    print('incorrect')
```

```python
email = "jhon78@gmail.com  jhon@.com david.989@yahoo.com"

print(re.findall("[\w._%]{0,20}@[\w-].[A-Za-z]{2,3}",email))
print(len(re.findall("[\w._%]{0,20}@[\w-].[A-Za-z]{2,3}",email)))
```

```python
import urllib.request
from re import findall
```

```python
url = "https://www.summet.com/dmsi/html/codesamples/addresses.html"

a = urllib.request.urlopen(url)
html = a.read()
htmlstr = html.decode()
phn = findall("\(\d{3}\) \d{3}-\d{4}",htmlstr)

for i in phn:
    print(i)
```
